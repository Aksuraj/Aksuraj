**Convert Datatype**


```python
data['Date'] = pd.to_datetime(data['Date'])
```

```python
data.info()
```

**Create month and year**

```python
data["Year"] = data["Date"].dt.year
data["Month"] = data["Date"].dt.month
```

```python
data.info()
```

**Create Seasons**

```python
def season(month):
  if month in [12,1,2]:
    return "Winter"

  elif month in [3,4,5]:
    return "Summer"

  elif month in [6,7,8,9]:
    return "Monsoon"

  else:
    return "Autumn"

data["Season"] = data["Month"].apply(season)
```

```python
data.head(10)
```

**Descriptive statistics**

```python
print("Mean =", np.mean(data["Temperature"]))
print("Median =", np.median(data["Temperature"]))
print("Standard Deviation =", np.std(data["Temperature"]))
```

```python
print("Mean =", np.mean(data["Humidity"]))
print("Median =", np.median(data["Humidity"]))
print("Standard Deviation =", np.std(data["Humidity"]))
```

```python
print("Mean =", np.mean(data["Precipitation"]))
print("Median =", np.median(data["Precipitation"]))
print("Standard Deviation =", np.std(data["Precipitation"]))
```

**Monthly Analysis**

```python
monthly = data.groupby("Month")[
    ["Temperature",
     "Humidity",
     "Precipitation"]
].mean()

monthly
```

**Seasonal Analysis**

```python
seasonal = data.groupby("Season")[
    ["Temperature",
     "Humidity",
     "Precipitation"]
].mean()

seasonal
```

**Temperature Trend**

```python
monthly_temp = data.groupby("Month")["Temperature"].mean()

plt.figure(figsize =(10,5))
plt.plot(
    monthly_temp.index,
    monthly_temp.values,
    marker = "o"
)
plt.title("Average Monthly Temperature")

plt.xlabel("Month")

plt.ylabel("Temperature")

plt.grid(True)

plt.show()
```

```python
monthly_precip = data.groupby("Month")["Precipitation"].mean()

plt.figure(figsize=(10,5))

plt.bar(
    monthly_precip.index,
    monthly_precip.values
)

plt.title("Average Monthly Precipitation")

plt.xlabel("Month")

plt.ylabel("Precipitation")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    data["Temperature"],
    bins=20,
    edgecolor="black"
)

plt.title("Temperature Distribution")

plt.xlabel("Temperature")

plt.ylabel("Frequency")

plt.show()
```
